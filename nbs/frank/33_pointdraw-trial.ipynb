## Earthsim example

Ok, this example works. I now see that the PointDraw instance takes the geoviews Points element as a source! Let's see if we can follow along the tutorial.

```python
import os
import numpy as np
import holoviews as hv
import geoviews as gv
import cartopy.crs as ccrs

from holoviews import opts
from holoviews.streams import PointDraw, PolyEdit, BoxEdit, PolyDraw, FreehandDraw

hv.extension('bokeh')

tiles = gv.tile_sources.ESRI.opts(width=900, height=500)
```

```python
points = gv.Points(np.random.rand(10, 2)*10)
point_stream = PointDraw(source=points, num_objects=10)

(tiles * points).opts(
    opts.Points(tools=['hover'], size=10, color='red'))
```

```python
point_stream.data
```

Let's work along the example here. 

```python
if point_stream.data:
    display(point_stream.element.dframe())
```

```python
sample_box = hv.Bounds((-90.99, 32.25, -90.85, 32.37))

box_poly = gv.Polygons([sample_box])
box_stream = BoxEdit(source=box_poly, num_objects=3)
(tiles * box_poly).opts(
    opts.Polygons(fill_alpha=0, line_color='white', selection_fill_color='red'))
```

```python
def bbox(poly):
    "Convert the polygon returned by the BoxEdit stream into a bounding box tuple"
    xs,ys = poly.array().T
    return (xs[0], ys[0], xs[2], ys[2])

if box_stream.element:
    polygons = box_stream.element.split()
    bboxes = [bbox(p) for p in polygons]
    print(bboxes)
```

```python
shapefile = './vicksburg_watershed/watershed_boundary.shp'
```

```python
mask_poly = gv.Shape.from_shapefile(shapefile)
vertex_stream = PolyEdit(source=mask_poly)

abc = tiles * mask_poly.opts(tools=['box_select'], alpha=0.5, line_width=3, color='white')
abc
```

```python
data = ([0, 0.5, 1], [0, 0.5, 0], ['red', 'green', 'blue'])
points = hv.Points(data, vdims='color').redim.range(x=(-.1, 1.1), y=(-.1, 1.1))
point_stream = streams.PointDraw(data=points.columns(), num_objects=10, source=points, empty_value='black')
table = hv.Table(points, ['x', 'y'], 'color')
DataLink(points, table)

(points + table).opts(
    opts.Layout(merge_tools=False),
    opts.Points(active_tools=['point_draw'], color='color', height=400,
                size=10, tools=['hover'], width=400),
    opts.Table(editable=True))
```
