Ok, this is going to work for our case. Next step is to create a function that reads the json rectangles data and returns a label list. Let's look up the json stuff.  

````python
import json

def get_labels(json_file='file.json'):
    '''Read rectangles data from  `json_file` and return corresponding `label_list`'''

    with open(json_file, 'r') as fh: 
        boxes_dict = json.load(fh) 

    label_list = [(x0, y0, f'#{i:02}') for i, [x0, y0] in enumerate(zip(boxes_dict['x0'], boxes_dict['y0']))]

    return label_list 
````

But this does not work as we need to supply a callable as an argument to a dynamic map: https://holoviews.org/getting_started/Live_Data.html#live-interaction 

A basic example is described here: https://stackoverflow.com/questions/62433507/update-holoviews-image-interactively-with-panel-widget

````python
# Source - https://stackoverflow.com/a
# Posted by philippjfr
# Retrieved 2025-12-19, License - CC BY-SA 4.0

import numpy as np
import holoviews as hv
import panel as pn 
from holoviews import opts

hv.extension('bokeh')

# Make grayscale test images
images = []
xmax, ymax = 2560, 1600
for i in range(5):
    # Gray background
    image = np.ones(shape=(ymax, xmax), dtype=np.uint8) * 200
    # Make each image unique with different position for a black square
    # with interior white square
    image[200*i:200*i + 400,200*i:200*i + 400] = 0
    image[200*i+175:200*i+175 + 50,200*i+175:200*i+175 + 50] = 255
    images.append(image)

# Make hv.Images objects
scale_factor = 8
bounds=(0, 0, xmax, ymax)   # Coordinate system: (left, bottom, right, top)
options = [
    opts.Image(cmap='gray',
               aspect='equal',
               frame_width=int(xmax/scale_factor),
               frame_height=int(ymax/scale_factor),
              )
]

# Set up selector object
image_selector = pn.widgets.MultiSelect(
    name="Choose image index", 
    options=[i for i in range(len(images))],
    size=6,
    width=120,
    value=[0, 1]
)

# Define interactivity
@pn.depends(indexes=image_selector)
def img0(indexes):
    index = indexes[0]
    return hv.Image(images[index], bounds=bounds).opts(*options)

@pn.depends(indexes=image_selector)
def img1(indexes):
    index = indexes[0]
    if index != len(images) - 1:
        data = images[index + 1]
    else:
        data = images[index]
    #return hv.Image(images[index], bounds=bounds).opts(*options)
    return hv.Image(data, bounds=bounds).opts(*options)


# Create panel layout
layout = pn.Row(
    image_selector,
    pn.Column(
        hv.DynamicMap(img0),
        hv.DynamicMap(img1)
    )
)
layout
````
