# How to tag

**This test shows that in principle it is possible to set custom tags (such as file hashes) to remote files. This would perhaps enable us to find files back even if they have been mistakenly renamed.**  

Also see this discussion: https://help.nextcloud.com/t/where-are-dav-properties-persisted-and-how-to-add-new-ones/81939/6

```python
from webdav3.client import Client
import os
import fairdatanow
```

Beware that webdavclient3 has another url. 

```python
options = {
    'webdav_hostname': "https://laboppad.nl/remote.php/dav/files/rce.fligterink", 
    'webdav_login':    os.getenv('NC_AUTH_USER'),
    'webdav_password': os.getenv('NC_AUTH_PASS')
}
```

```python
client = Client(options)
```

```python
test_path = 'falnama-project/OPENDATA/xrays.html'
```

```python
client.list(remote_path='falnama-project/OPENDATA/', get_info=True)
```

```python
client.set_property(test_path, option={'name':'test', 'value': 'abc'})
```

```python
client.get_property(test_path, option={'name': 'test'})
```
