## Our first holoviews peak pattern atlas 

**Let's see if our mysterious disappearing-curves bug stays away...**

No, the problem is back. How to fix it??? Answer: dump firefox and use chromium instead.

```python
from moseley import ElementXRF, PeakPatternAtlas 
import numpy as np 
import holoviews as hv 
```

```python
hv.extension('bokeh')
```

Need faster access to elements of interest.

```python
ppa = PeakPatternAtlas()
```

```python
eoi = ppa.EOI
```

```python
scatters = []
lines = []
yticks = []
energy = hv.Dimension('Energy', unit='keV')

for i, element in enumerate(eoi): 
    xrf = ElementXRF(element)
    peaks_x, peaks_y = xrf.peaks_xy.T 
    x_min = float(peaks_x.min()) 
    x_max =float(peaks_x.max())
    ptrn_y = np.ones_like(peaks_y) * i
    
    scatters.append(hv.Scatter((peaks_x, ptrn_y), energy, 'Element').opts(size=7))
    lines.append(hv.Curve([(x_min, i), (x_max, i)], energy, 'Element'))

    yticks.append((i, element))

atlas = hv.Overlay(lines) * hv.Overlay(scatters)
atlas = atlas.opts(yticks=yticks, frame_height=500, frame_width=500, xlim=(0, 22))

```

```python
from fairdatanow import DataViewer
import os
import numpy as np 
import holoviews as hv
from holoviews import opts
from holoviews.operation.datashader import rasterize
import re 
import maxrf4u as mx 
import panel as pn 
```

```python
hv.extension('bokeh')
```

```python
#|code-fold: true 
configuration = {
    'url': "https://laboppad.nl/falnama-project", 
    'user':    os.getenv('NC_AUTH_USER'),
    'password': os.getenv('NC_AUTH_PASS')
}
```

```python
filters = {'extensions': ['.datastack']} 

dv = DataViewer(configuration, **filters)
datastack_files = dv.download_filtered() 
```

```python
object_nums = [re.sub(r'.*(WM-71803-\d\d).*', r'\1', dsf) for dsf in datastack_files]
```

```python
n = 15
datastack_file = datastack_files[n]
ds = mx.DataStack(datastack_file)

object_num = object_nums[n]

cube = ds.read('maxrf_cube', compute=False)
imvis_highres = ds.read('imvis_reg_highres')
imvis = ds.read('imvis_reg')
extent = ds.read('imvis_extent')
```

```python
h, w, d = imvis.shape
```

```python
image = rasterize(hv.RGB(imvis_highres, bounds=[-0.5,h-0.5,w-0.5,-0.5]), height=4000, width=4000)
image2 = hv.RGB(imvis, bounds=[-0.5,h-0.5,w-0.5,-0.5])
roi = image[60:130,170:230]
points = hv.Points((85,182))
roi = roi.redim(x='detail_x', y='detail_y');
```

```python
layout = image * points + roi * points + image2 * points
layout = layout.opts(
    opts.Points(marker='+', size=15, color='r', axiswise=True),
    opts.RGB(aspect='equal',axiswise=True),
    opts.Layout(toolbar='left')
            ).cols(2)
```
