# This has been a roller coaster 

*Looking back on our Falnama project coding adventure what did we set out to do? Which problems did we encounter? How did we solve them? And how to move forward from here on?* 

## Towards building an interactive data dashboard 

The main objective of our coding efforts in the context of the Falnama project seemed pretty straightforward: build a Jupyter based data dashboard to enable better interdisciplinary research in the Falnama project and preparing for an interactive website for the Wereldmuseum. It was agreed within our team the first priority was to visualize the MA-XRF data for sixteen scans that had been acquired earlier this year. At the time this seemed to us a quite manageable objective...   

On the road however, it turned out that we had to resolve a rather large number of interesting but tough technical issues in order to automate the rendering of the MA-XRF data into a legible form. The development process has been quite a roller coaster. And this is a good time to look back to review our progress and decide how to consolidate it and move forward from here.
