```python
import nc_py_api
import os
```

```python
configuration = {
    'nextcloud_url': "https://laboppad.nl/",
    'nc_auth_user': os.getenv("NC_AUTH_USER"),
    'nc_auth_pass': os.getenv("NC_AUTH_PASS")
}

nc_py_api.options.NPA_NC_CERT = False
```

```python
nc = nc_py_api.Nextcloud(**configuration)
```

```python
files = nc.files.listdir("falnama-project", -1)
```

```python
for file in files:
    if file.name.endswith('raw') and not file.is_dir:
        print(file.name)
```
