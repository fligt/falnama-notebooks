```python
import holoviews as hv
from holoviews.operation.datashader import rasterize 
from holonote.annotate import Annotator
import panel as pn
from holonote.app import PanelWidgets
import imageio.v3 as iio
import numpy as np
hv.extension('bokeh')
```

# Load the plot

```python
points = [(0.1*i, np.sin(0.1*i)) for i in range(100)]
curve = hv.Curve(points)
```

```python
curve
```

# Add the annotator

```python
annotator = Annotator(curve, fields=['description'])
annotator * curve
```

# Commit descriptions

```python
annotator.add_annotation(description='Mijn annotatie')
annotator.commit()
```

# Build an app

```python
pn.Row(PanelWidgets(annotator), annotator * curve)
```
