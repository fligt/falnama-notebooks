```python
import falnama
import matplotlib.pyplot as plt
import maxrf4u as mx
import re
import os
import numpy as np
import skimage.transform as skt
```

```python
def get_object_name(datastack_file):
    return re.sub(r'.*(71803-\d\d).*', r'\1', datastack_file)
```

```python
os.getcwd()
```

```python
datastack_files = falnama.Falnama_maxrf_data().get_datastack_files()
```

```python
norm_dict = {'S': 4.537112712860107, 'Ag': 16.644882202148438, 'K': 107.45109558105469, 'Ca': 87.45025634765625, 'Ba': 19.0678653717041, 'Ti': 31.38340950012207, 'Cr': 139.6975555419922, 'Mn': 210.4607391357422, 'Fe': 190.41091918945312, 'Ni': 2645.46875, 'Cu': 184.11544799804688, 'Zn': 20.21651840209961, 'Au': 114.6514892578125, 'Hg': 114.73220825195312, 'As': 212.54464721679688, 'Pb': 111.24705505371094, 'Sr': 26.372350692749023, 'Zr': 25.659061431884766}
```

```python
for datastack_file in datastack_files: 
    
    objectnr = get_object_name(datastack_file)
    folder = f'{objectnr}_upscaled'
    os.mkdir(folder)
    
    ds = mx.DataStack(datastack_file)
    imvis = ds.read('imvis_reg_highres')
    element_maps = ds.read('nmf_elementmaps')
    atom_nums = ds.read('nmf_atomnums')
    elements = mx.elems_from_atomnums(atom_nums)

    h, w, d = imvis.shape

    plt.imsave(f'{folder}/rgb.jpg', imvis, dpi=500)

    for i, elem in enumerate(norm_dict):
        vmax = norm_dict[elem]
        print(f'{i}/{len(norm_dict)-1}', end='\r')
        upsized_map = skt.resize(element_maps[i], [h, w])
        plt.imsave(f'{folder}/{elem}.jpg', upsized_map, dpi=500, vmin=0, vmax=vmax)
```
