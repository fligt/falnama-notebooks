## Interacting with JSON-files

Let's start by initializing a file. The first step is to import the json library

```python
import json
```

```python
x = {
    "datastackfile": None,
    "rois": []
}

y = json.dumps(x, indent=4)

print(y)
```

### Change value

```python
x = json.loads(y)
x['datastackfile'] = 'path/to/datastack'

y = json.dumps(x, indent=4)
print(y)
```

### Append

```python
x = json.loads(y)
x['rois'].append({'coords':[50,100,150,200],
                 'elements':[]})
                 

y = json.dumps(x, indent=4)
print(y)
```

### Remove

```python
x = json.loads(y)
x['rois'].pop(0)

y = json.dumps(x, indent=4)
print(y)
```
