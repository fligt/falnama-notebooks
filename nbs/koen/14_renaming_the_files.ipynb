```python
from fairdatanow import RemoteData2
import os
import panel as pn
import re
```

```python
configuration = {
    'url': "https://laboppad.nl/falnama-project",
    'user': os.getenv("NC_AUTH_USER"),
    'password': os.getenv("NC_AUTH_PASS")
}
```

```python
rm = RemoteData2(configuration)
```

```python
file_table = rm.listdir("falnama-project")
```

```python
file_table
```

```python
xray_list = rm.file_table.selected_dataframe['path'].tolist()
xray_list
```

```python
def rename_xrays(xray_list):
    pattern = r'(\d{5}-\d+)_(\d{4}-\d{2}-\d{2})'
    new_paths = []
    for path in xray_list:
        # split the file_name from the full path
        location, file_name = os.path.split(path)
        # try and match the file_name to the regex pattern defined
        match = re.search(pattern, file_name)
        # if match is found save the groups
        if match:
            object_nr = match.group(1)
            date = match.group(2)
        else:
            print(f"One file could not be matched: {file_name}")
            return

        new_paths.append(f"{location}/{object_nr}_RMA_xrays_{date}.zip")

    return new_paths
        
        
```

```python
rename_xrays(xray_list)
```
